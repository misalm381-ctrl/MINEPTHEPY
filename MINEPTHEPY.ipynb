```python
# =====================================================================
# MINEPTHEPY
# =====================================================================

import os
import re
import zipfile
import urllib.request
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error

warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")

PROJECT_NAME = "MINEPTHEPY"
DATA_DIR = "/content/ml-latest-small"
OUTPUT_DIR = "/content/minepthepy_outputs"

os.makedirs(OUTPUT_DIR, exist_ok=True)

# =====================================================================
# SECTION 1: DATASET LOADING
# =====================================================================

print("=" * 75)
print("                 MINEPTHEPY MOVIE RECOMMENDATION SYSTEM")
print("=" * 75)

print("\nSECTION 1: LOADING DATASET")
print("-" * 75)


def load_dataset():

    movies_path = os.path.join(DATA_DIR, "movies.csv")
    ratings_path = os.path.join(DATA_DIR, "ratings.csv")
    tags_path = os.path.join(DATA_DIR, "tags.csv")
    links_path = os.path.join(DATA_DIR, "links.csv")

    if not os.path.exists(movies_path) or not os.path.exists(ratings_path):

        print("MovieLens dataset not found. Downloading...")

        zip_path = "/content/ml-latest-small.zip"

        url = (
            "https://files.grouplens.org/datasets/movielens/"
            "ml-latest-small.zip"
        )

        urllib.request.urlretrieve(url, zip_path)

        with zipfile.ZipFile(zip_path, "r") as archive:
            archive.extractall("/content")

        print("Dataset downloaded successfully.")

    movies = pd.read_csv(movies_path)
    ratings = pd.read_csv(ratings_path)

    tags = (
        pd.read_csv(tags_path)
        if os.path.exists(tags_path)
        else pd.DataFrame(
            columns=["userId", "movieId", "tag", "timestamp"]
        )
    )

    links = (
        pd.read_csv(links_path)
        if os.path.exists(links_path)
        else pd.DataFrame(
            columns=["movieId", "imdbId", "tmdbId"]
        )
    )

    return movies, ratings, tags, links


raw_movies, raw_ratings, raw_tags, raw_links = load_dataset()

print("Dataset loaded successfully.")

# Keep original data for the cleaning report.
movies = raw_movies.copy()
ratings = raw_ratings.copy()
tags = raw_tags.copy()
links = raw_links.copy()

# =====================================================================
# SECTION 2: INITIAL DATASET INFORMATION
# =====================================================================

print("\nSECTION 2: ORIGINAL DATASET INFORMATION")
print("-" * 75)

original_summary = pd.DataFrame({
    "Dataset": ["Movies", "Ratings", "Tags", "Links"],
    "Original Rows": [
        len(raw_movies),
        len(raw_ratings),
        len(raw_tags),
        len(raw_links)
    ],
    "Original Columns": [
        len(raw_movies.columns),
        len(raw_ratings.columns),
        len(raw_tags.columns),
        len(raw_links.columns)
    ],
    "Memory Usage (MB)": [
        round(raw_movies.memory_usage(deep=True).sum() / 1048576, 3),
        round(raw_ratings.memory_usage(deep=True).sum() / 1048576, 3),
        round(raw_tags.memory_usage(deep=True).sum() / 1048576, 3),
        round(raw_links.memory_usage(deep=True).sum() / 1048576, 3)
    ]
})

display(original_summary)

print("\nMOVIES DATASET PREVIEW")
display(raw_movies.head())

print("\nRATINGS DATASET PREVIEW")
display(raw_ratings.head())

print("\nTAGS DATASET PREVIEW")
display(raw_tags.head())

print("\nLINKS DATASET PREVIEW")
display(raw_links.head())

# =====================================================================
# SECTION 3: DATA CLEANING
# =====================================================================

print("\nSECTION 3: DATA CLEANING")
print("-" * 75)

cleaning_log = []


def record_cleaning(dataset_name, step, before, after, explanation):

    cleaning_log.append({
        "Dataset": dataset_name,
        "Cleaning Operation": step,
        "Rows Before": before,
        "Rows After": after,
        "Rows Removed": before - after,
        "Explanation": explanation
    })


# ---------------- MOVIES CLEANING ----------------

before = len(movies)

movies = movies.dropna(subset=["movieId", "title"])
record_cleaning(
    "Movies",
    "Remove missing movie IDs or titles",
    before,
    len(movies),
    "Rows without essential movie information are removed."
)

before = len(movies)

movies["movieId"] = pd.to_numeric(
    movies["movieId"], errors="coerce"
)

movies = movies.dropna(subset=["movieId"])

movies["movieId"] = movies["movieId"].astype(int)

record_cleaning(
    "Movies",
    "Validate movie IDs",
    before,
    len(movies),
    "Movie IDs are converted to valid integers."
)

before = len(movies)

movies["title"] = movies["title"].astype(str).str.strip()
movies["genres"] = movies["genres"].fillna("").astype(str).str.strip()

movies = movies[
    movies["title"].ne("")
    & movies["title"].ne("nan")
]

record_cleaning(
    "Movies",
    "Clean movie titles and genres",
    before,
    len(movies),
    "Whitespace is removed and missing genres are represented as empty."
)

before = len(movies)

movies = movies.drop_duplicates(subset=["movieId"], keep="first")

record_cleaning(
    "Movies",
    "Remove duplicate movie IDs",
    before,
    len(movies),
    "Only one record per movie ID is retained."
)


# ---------------- RATINGS CLEANING ----------------

before = len(ratings)

ratings = ratings.dropna(
    subset=["userId", "movieId", "rating"]
)

record_cleaning(
    "Ratings",
    "Remove missing essential values",
    before,
    len(ratings),
    "Rows missing user ID, movie ID, or rating are removed."
)

before = len(ratings)

ratings["userId"] = pd.to_numeric(
    ratings["userId"], errors="coerce"
)

ratings["movieId"] = pd.to_numeric(
    ratings["movieId"], errors="coerce"
)

ratings["rating"] = pd.to_numeric(
    ratings["rating"], errors="coerce"
)

ratings = ratings.dropna(
    subset=["userId", "movieId", "rating"]
)

record_cleaning(
    "Ratings",
    "Validate numeric values",
    before,
    len(ratings),
    "Invalid user IDs, movie IDs, and ratings are removed."
)

ratings["userId"] = ratings["userId"].astype(int)
ratings["movieId"] = ratings["movieId"].astype(int)

before = len(ratings)

ratings = ratings[
    ratings["rating"].between(0.5, 5.0)
]

record_cleaning(
    "Ratings",
    "Validate rating range",
    before,
    len(ratings),
    "Only ratings between 0.5 and 5.0 are retained."
)

before = len(ratings)

ratings = ratings.drop_duplicates(
    subset=["userId", "movieId"],
    keep="last"
)

record_cleaning(
    "Ratings",
    "Remove duplicate user-movie pairs",
    before,
    len(ratings),
    "Only one rating per user-movie pair is retained."
)

# Keep only ratings for movies present in the cleaned movie table.
before = len(ratings)

ratings = ratings[
    ratings["movieId"].isin(movies["movieId"])
].copy()

record_cleaning(
    "Ratings",
    "Remove ratings for unknown movies",
    before,
    len(ratings),
    "Ratings must refer to a movie in the movies dataset."
)


# ---------------- TAGS CLEANING ----------------

if not tags.empty:

    before = len(tags)

    tags = tags.dropna(subset=["movieId", "tag"])

    record_cleaning(
        "Tags",
        "Remove missing movie IDs and tags",
        before,
        len(tags),
        "Rows missing essential tag information are removed."
    )

    before = len(tags)

    tags["movieId"] = pd.to_numeric(
        tags["movieId"], errors="coerce"
    )

    tags = tags.dropna(subset=["movieId"])

    tags["movieId"] = tags["movieId"].astype(int)
    tags["tag"] = tags["tag"].astype(str).str.strip()

    tags = tags[tags["tag"] != ""]

    record_cleaning(
        "Tags",
        "Validate and clean tags",
        before,
        len(tags),
        "Invalid movie IDs and empty tags are removed."
    )

    before = len(tags)

    tags = tags.drop_duplicates()

    record_cleaning(
        "Tags",
        "Remove exact duplicate tag records",
        before,
        len(tags),
        "Exact duplicate tag rows are removed."
    )

    before = len(tags)

    tags = tags[tags["movieId"].isin(movies["movieId"])]

    record_cleaning(
        "Tags",
        "Remove tags for unknown movies",
        before,
        len(tags),
        "Tags must reference an existing movie."
    )


# ---------------- LINKS CLEANING ----------------

if not links.empty and "movieId" in links.columns:

    before = len(links)

    links = links.dropna(subset=["movieId"])

    links["movieId"] = pd.to_numeric(
        links["movieId"], errors="coerce"
    )

    links = links.dropna(subset=["movieId"])

    links["movieId"] = links["movieId"].astype(int)

    record_cleaning(
        "Links",
        "Validate movie IDs",
        before,
        len(links),
        "Invalid movie IDs are removed."
    )

    before = len(links)

    links = links.drop_duplicates(
        subset=["movieId"],
        keep="first"
    )

    record_cleaning(
        "Links",
        "Remove duplicate movie IDs",
        before,
        len(links),
        "Only one external link record per movie is retained."
    )


print("\nDATA CLEANING PROCESS COMPLETED")

cleaning_report = pd.DataFrame(cleaning_log)

display(cleaning_report)

# =====================================================================
# SECTION 4: MISSING VALUES AND DUPLICATE ANALYSIS
# =====================================================================

print("\nSECTION 4: MISSING VALUES ANALYSIS")
print("-" * 75)


def missing_report(df, dataset_name):

    report = pd.DataFrame({
        "Column": df.columns,
        "Missing Values": [
            df[col].isna().sum() for col in df.columns
        ],
        "Missing Percentage (%)": [
            round(df[col].isna().mean() * 100, 2)
            for col in df.columns
        ]
    })

    report.insert(0, "Dataset", dataset_name)

    return report


missing_values_report = pd.concat([
    missing_report(movies, "Movies"),
    missing_report(ratings, "Ratings"),
    missing_report(tags, "Tags"),
    missing_report(links, "Links")
], ignore_index=True)

display(missing_values_report)

print("\nDUPLICATE RECORD ANALYSIS")

duplicate_report = pd.DataFrame({
    "Dataset": ["Movies", "Ratings", "Tags", "Links"],

    "Exact Duplicate Rows": [
        raw_movies.duplicated().sum(),
        raw_ratings.duplicated().sum(),
        raw_tags.duplicated().sum(),
        raw_links.duplicated().sum()
    ],

    "Final Rows After Cleaning": [
        len(movies),
        len(ratings),
        len(tags),
        len(links)
    ]
})

display(duplicate_report)

# =====================================================================
# SECTION 5: DATA PROCESSING AND FEATURE ENGINEERING
# =====================================================================

print("\nSECTION 5: DATA PROCESSING")
print("-" * 75)

movies["release_year"] = pd.to_numeric(
    movies["title"].str.extract(r"\((\d{4})\)\s*$")[0],
    errors="coerce"
)

movies["genres_list"] = movies["genres"].apply(
    lambda value: [
        genre for genre in str(value).split("|")
        if genre and genre != "(no genres listed)"
    ]
)

if not tags.empty:

    tag_features = tags.groupby("movieId")["tag"].apply(
        lambda values: " ".join(values.astype(str))
    ).to_dict()

else:

    tag_features = {}

movies["tag_features"] = movies["movieId"].map(
    tag_features
).fillna("")

movies["combined_features"] = (
    movies["genres"].str.replace("|", " ", regex=False)
    + " "
    + movies["tag_features"]
).str.strip()

movie_rating_stats = ratings.groupby("movieId").agg(
    average_rating=("rating", "mean"),
    rating_count=("rating", "count"),
    rating_std=("rating", "std")
).reset_index()

movies = movies.merge(
    movie_rating_stats,
    on="movieId",
    how="left"
)

movies["rating_count"] = movies["rating_count"].fillna(0).astype(int)

movies["average_rating"] = movies["average_rating"].fillna(0)

movies["rating_std"] = movies["rating_std"].fillna(0)

print("Feature engineering completed.")
print("Features created: release year, genres list, tag features,")
print("average rating, rating count, and rating standard deviation.")

print("\nPROCESSED DATA PREVIEW")

display(
    movies[
        [
            "movieId",
            "title",
            "genres",
            "release_year",
            "average_rating",
            "rating_count"
        ]
    ].head(15)
)

# =====================================================================
# SECTION 6: DATASET SUMMARY
# =====================================================================

print("\nSECTION 6: FINAL DATASET SUMMARY")
print("-" * 75)

all_genres = sorted({
    genre
    for genre_list in movies["genres_list"]
    for genre in genre_list
})

summary = pd.DataFrame({
    "Metric": [
        "Movies Before Cleaning",
        "Movies After Cleaning",
        "Movie Records Removed",
        "Ratings Before Cleaning",
        "Ratings After Cleaning",
        "Rating Records Removed",
        "Unique Users",
        "Unique Genres",
        "Tag Records After Cleaning",
        "External Link Records",
        "Average Rating",
        "Median Rating",
        "Highest Rating",
        "Lowest Rating",
        "Total Rating Interactions",
        "Rating Sparsity (%)"
    ],

    "Value": [
        len(raw_movies),
        len(movies),
        len(raw_movies) - len(movies),
        len(raw_ratings),
        len(ratings),
        len(raw_ratings) - len(ratings),
        ratings["userId"].nunique(),
        len(all_genres),
        len(tags),
        len(links),
        round(ratings["rating"].mean(), 3),
        round(ratings["rating"].median(), 3),
        ratings["rating"].max(),
        ratings["rating"].min(),
        len(ratings),
        round(
            (
                1 - len(ratings) /
                max(
                    ratings["userId"].nunique()
                    * movies["movieId"].nunique(),
                    1
                )
            ) * 100,
            2
        )
    ]
})

display(summary)

# =====================================================================
# SECTION 7: SORTED MOVIES AND POPULARITY ANALYSIS
# =====================================================================

print("\nSECTION 7: MOVIE RANKINGS")
print("-" * 75)


def calculate_popular_movies(min_ratings=10):

    stats = ratings.groupby("movieId").agg(
        average_rating=("rating", "mean"),
        rating_count=("rating", "count")
    ).reset_index()

    global_mean = ratings["rating"].mean()

    stats["weighted_rating"] = (
        (
            stats["rating_count"]
            / (stats["rating_count"] + min_ratings)
        ) * stats["average_rating"]
        +
        (
            min_ratings
            / (stats["rating_count"] + min_ratings)
        ) * global_mean
    )

    result = movies.drop(
        columns=[
            "average_rating",
            "rating_count",
            "rating_std"
        ],
        errors="ignore"
    ).merge(
        stats,
        on="movieId",
        how="left"
    )

    result["average_rating"] = result["average_rating"].fillna(0)
    result["rating_count"] = result["rating_count"].fillna(0)
    result["weighted_rating"] = result["weighted_rating"].fillna(0)

    return result


popular_movies = calculate_popular_movies()

print("\nTOP 20 MOST-RATED MOVIES")

most_rated = popular_movies.sort_values(
    "rating_count",
    ascending=False
).head(20)

display(
    most_rated[
        ["title", "genres", "rating_count", "average_rating"]
    ].reset_index(drop=True)
)

print("\nTOP 20 POPULAR MOVIES")
print("Ranked using a weighted rating to account for rating count.")

display(
    popular_movies.sort_values(
        "weighted_rating",
        ascending=False
    )[
        [
            "title",
            "genres",
            "average_rating",
            "rating_count",
            "weighted_rating"
        ]
    ].head(20).reset_index(drop=True)
)

print("\nTOP 20 HIGHEST-RATED MOVIES")
print("Only movies with at least 50 ratings are included.")

top_rated = popular_movies[
    popular_movies["rating_count"] >= 50
].sort_values(
    "average_rating",
    ascending=False
).head(20)

display(
    top_rated[
        ["title", "genres", "average_rating", "rating_count"]
    ].reset_index(drop=True)
)

print("\nMOVIES SORTED ALPHABETICALLY")

display(
    movies.sort_values("title")[
        ["movieId", "title", "genres", "release_year"]
    ].head(30).reset_index(drop=True)
)

# =====================================================================
# SECTION 8: EXPLORATORY DATA ANALYSIS AND VISUALIZATIONS
# =====================================================================

print("\nSECTION 8: EXPLORATORY DATA ANALYSIS")
print("-" * 75)

# ---------------- GRAPH 1: RATING DISTRIBUTION ----------------

plt.figure(figsize=(10, 5))

sns.countplot(
    data=ratings,
    x="rating"
)

plt.title("Distribution of Movie Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Ratings")
plt.tight_layout()
plt.show()

# ---------------- GRAPH 2: MOVIE GENRES ----------------

genre_counts = pd.Series(
    [
        genre
        for genre_list in movies["genres_list"]
        for genre in genre_list
    ]
).value_counts()

print("\nMOVIE COUNTS BY GENRE")

display(
    genre_counts.rename_axis("Genre")
    .reset_index(name="Number of Movies")
)

plt.figure(figsize=(12, 7))

sns.barplot(
    x=genre_counts.values,
    y=genre_counts.index
)

plt.title("Number of Movies by Genre")
plt.xlabel("Number of Movies")
plt.ylabel("Genre")
plt.tight_layout()
plt.show()

# ---------------- GRAPH 3: RELEASE YEAR ----------------

release_year_counts = (
    movies.dropna(subset=["release_year"])
    .groupby("release_year")
    .size()
    .sort_index()
)

print("\nMOVIES BY RELEASE YEAR")

display(
    release_year_counts.rename_axis("Release Year")
    .reset_index(name="Number of Movies")
    .tail(20)
)

plt.figure(figsize=(13, 5))

plt.plot(
    release_year_counts.index,
    release_year_counts.values
)

plt.title("Movie Release Trend by Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Movies")
plt.tight_layout()
plt.show()

# ---------------- GRAPH 4: MOST-RATED MOVIES ----------------

top_most_rated = popular_movies.sort_values(
    "rating_count",
    ascending=False
).head(15)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=top_most_rated,
    x="rating_count",
    y="title"
)

plt.title("15 Most-Rated Movies")
plt.xlabel("Number of Ratings")
plt.ylabel("Movie")
plt.tight_layout()
plt.show()

# ---------------- GRAPH 5: TOP-RATED MOVIES ----------------

top_rating_graph = popular_movies[
    popular_movies["rating_count"] >= 50
].sort_values(
    "average_rating",
    ascending=False
).head(15)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=top_rating_graph,
    x="average_rating",
    y="title"
)

plt.title("15 Highest-Rated Movies with at Least 50 Ratings")
plt.xlabel("Average Rating")
plt.ylabel("Movie")
plt.xlim(0, 5)
plt.tight_layout()
plt.show()

# ---------------- GRAPH 6: RATINGS PER USER ----------------

ratings_per_user = ratings.groupby("userId").size()

plt.figure(figsize=(10, 5))

plt.hist(
    ratings_per_user,
    bins=30
)

plt.title("Distribution of Ratings per User")
plt.xlabel("Number of Ratings Given by a User")
plt.ylabel("Number of Users")
plt.tight_layout()
plt.show()

# ---------------- GRAPH 7: RATINGS PER MOVIE ----------------

ratings_per_movie = ratings.groupby("movieId").size()

plt.figure(figsize=(10, 5))

plt.hist(
    ratings_per_movie,
    bins=30
)

plt.title("Distribution of Ratings per Movie")
plt.xlabel("Number of Ratings Received by a Movie")
plt.ylabel("Number of Movies")
plt.tight_layout()
plt.show()

# ---------------- GRAPH 8: AVERAGE RATING BY GENRE ----------------

genre_rating_rows = movies[
    ["movieId", "genres_list"]
].explode("genres_list")

genre_rating_rows = genre_rating_rows.rename(
    columns={"genres_list": "genre"}
)

genre_rating_rows = genre_rating_rows[
    genre_rating_rows["genre"].notna()
    & (genre_rating_rows["genre"] != "")
]

genre_rating_rows = genre_rating_rows.merge(
    ratings[["movieId", "rating"]],
    on="movieId",
    how="inner"
)

genre_average_rating = genre_rating_rows.groupby(
    "genre"
)["rating"].mean().sort_values(ascending=False)

print("\nAVERAGE USER RATING BY GENRE")

display(
    genre_average_rating.rename_axis("Genre")
    .reset_index(name="Average Rating")
    .round(3)
)

plt.figure(figsize=(12, 7))

sns.barplot(
    x=genre_average_rating.values,
    y=genre_average_rating.index
)

plt.title("Average Rating by Genre")
plt.xlabel("Average Rating")
plt.ylabel("Genre")
plt.xlim(0, 5)
plt.tight_layout()
plt.show()

# =====================================================================
# SECTION 9: PREPARE RECOMMENDATION MODELS
# =====================================================================

print("\nSECTION 9: PREPARING RECOMMENDATION MODELS")
print("-" * 75)

# Build content features.

vectorizer = TfidfVectorizer(
    stop_words="english",
    max_features=10000
)

tfidf_matrix = vectorizer.fit_transform(
    movies["combined_features"].fillna("")
)

# Movie-to-row index lookup.

movie_index_map = pd.Series(
    movies.index,
    index=movies["movieId"]
).to_dict()

# Train/test split.

if len(ratings) >= 5:

    train_ratings, test_ratings = train_test_split(
        ratings,
        test_size=0.20,
        random_state=42
    )

else:

    train_ratings = ratings.copy()
    test_ratings = ratings.copy()

train_matrix = train_ratings.pivot_table(
    index="userId",
    columns="movieId",
    values="rating",
    aggfunc="mean"
)

user_item_filled = train_matrix.fillna(0)

user_similarity_values = cosine_similarity(
    user_item_filled
)

user_similarity_df = pd.DataFrame(
    user_similarity_values,
    index=user_item_filled.index,
    columns=user_item_filled.index
)

print("Content-based model prepared.")
print("User similarity matrix prepared.")
print("Training and testing datasets prepared.")
print("All models are ready.")

# =====================================================================
# SECTION 10: RECOMMENDATION FUNCTIONS
# =====================================================================

def recommend_similar_movies(movie_title, top_n=10):

    matches = movies[
        movies["title"].str.contains(
            movie_title,
            case=False,
            regex=False,
            na=False
        )
    ]

    if matches.empty:

        print("\nNo matching movie found.")
        return

    print("\nMATCHING MOVIES")

    display(
        matches[
            ["movieId", "title", "genres", "average_rating"]
        ].head(15).reset_index(drop=True)
    )

    selected = matches.iloc[0]

    movie_id = int(selected["movieId"])
    position = movie_index_map[movie_id]

    scores = cosine_similarity(
        tfidf_matrix[position],
        tfidf_matrix
    ).flatten()

    ranked_positions = np.argsort(scores)[::-1]

    results = []

    for pos in ranked_positions:

        candidate = movies.iloc[pos]

        if int(candidate["movieId"]) == movie_id:
            continue

        results.append({
            "Movie": candidate["title"],
            "Genres": candidate["genres"],
            "Average Rating": round(
                float(candidate["average_rating"]), 2
            ),
            "Similarity Score": round(
                float(scores[pos]), 4
            )
        })

        if len(results) >= top_n:
            break

    print("\nRECOMMENDATIONS SIMILAR TO:", selected["title"])

    display(pd.DataFrame(results))


def recommend_for_user(user_id, top_n=10):

    user_id = int(user_id)

    if user_id not in train_matrix.index:

        print("\nUser ID not found in the training data.")
        print(
            "Example valid user IDs:",
            sorted(ratings["userId"].unique())[:20]
        )
        return

    similarities = user_similarity_df[user_id].drop(
        labels=[user_id],
        errors="ignore"
    )

    similarities = similarities[
        similarities > 0
    ].sort_values(ascending=False).head(30)

    if similarities.empty:

        print("Not enough similar users to make recommendations.")
        return

    user_ratings = train_matrix.loc[user_id]

    already_rated = user_ratings[
        user_ratings.notna()
    ].index

    candidate_ids = train_matrix.columns[
        ~train_matrix.columns.isin(already_rated)
    ]

    recommendations = []

    for movie_id in candidate_ids:

        neighbor_ratings = train_matrix.loc[
            similarities.index, movie_id
        ]

        valid = neighbor_ratings.notna()

        if not valid.any():
            continue

        selected_ratings = neighbor_ratings[valid]

        weights = similarities.reindex(
            selected_ratings.index
        ).fillna(0)

        if weights.sum() <= 0:
            continue

        predicted_rating = np.average(
            selected_ratings.values,
            weights=weights.values
        )

        recommendations.append({
            "movieId": int(movie_id),
            "Predicted Rating": round(
                float(predicted_rating), 3
            )
        })

    if not recommendations:

        print("No recommendations could be generated.")
        return

    result = pd.DataFrame(recommendations).sort_values(
        "Predicted Rating",
        ascending=False
    ).head(top_n)

    result = result.merge(
        movies[["movieId", "title", "genres"]],
        on="movieId",
        how="left"
    )

    result = result[
        [
            "title",
            "genres",
            "Predicted Rating"
        ]
    ]

    result.columns = [
        "Recommended Movie",
        "Genres",
        "Predicted Rating"
    ]

    print(f"\nPERSONALIZED RECOMMENDATIONS FOR USER {user_id}")

    display(result.reset_index(drop=True))


def search_movies():

    keyword = input(
        "\nEnter a movie title or keyword: "
    ).strip()

    if not keyword:
        print("Search keyword cannot be empty.")
        return

    results = movies[
        movies["title"].str.contains(
            keyword,
            case=False,
            regex=False,
            na=False
        )
    ]

    if results.empty:

        print("No movies found.")
        return

    print(f"\nFound {len(results)} movies.")

    display(
        results[
            [
                "movieId",
                "title",
                "genres",
                "release_year",
                "average_rating",
                "rating_count"
            ]
        ].head(50).reset_index(drop=True)
    )


def show_popular_movies():

    number = input(
        "\nNumber of movies to display (default 20): "
    ).strip()

    number = int(number) if number.isdigit() else 20

    number = max(1, min(number, len(movies)))

    result = calculate_popular_movies().sort_values(
        "weighted_rating",
        ascending=False
    ).head(number)

    print("\nPOPULAR MOVIES")

    display(
        result[
            [
                "title",
                "genres",
                "average_rating",
                "rating_count",
                "weighted_rating"
            ]
        ].reset_index(drop=True)
    )

    plt.figure(figsize=(12, 7))

    sns.barplot(
        data=result.head(15),
        x="weighted_rating",
        y="title"
    )

    plt.title("Popular Movies by Weighted Rating")
    plt.xlabel("Weighted Rating")
    plt.ylabel("Movie")
    plt.tight_layout()
    plt.show()


def explore_genres():

    print("\nGENRE DISTRIBUTION")

    display(
        genre_counts.rename_axis("Genre")
        .reset_index(name="Number of Movies")
    )

    plt.figure(figsize=(12, 7))

    sns.barplot(
        x=genre_counts.values,
        y=genre_counts.index
    )

    plt.title("Movies by Genre")
    plt.xlabel("Number of Movies")
    plt.ylabel("Genre")
    plt.tight_layout()
    plt.show()

    print("\nAVERAGE RATING BY GENRE")

    display(
        genre_average_rating.rename_axis("Genre")
        .reset_index(name="Average Rating")
        .round(3)
    )

    plt.figure(figsize=(12, 7))

    sns.barplot(
        x=genre_average_rating.values,
        y=genre_average_rating.index
    )

    plt.title("Average Rating by Genre")
    plt.xlabel("Average Rating")
    plt.ylabel("Genre")
    plt.xlim(0, 5)
    plt.tight_layout()
    plt.show()


def show_dataset_statistics():

    print("\nDATASET SUMMARY")

    display(summary)

    print("\nDATA CLEANING REPORT")

    display(cleaning_report)

    print("\nMISSING VALUE REPORT")

    display(missing_values_report)

    print("\nDUPLICATE RECORD REPORT")

    display(duplicate_report)

    print("\nTOP 20 MOST-RATED MOVIES")

    display(
        popular_movies.sort_values(
            "rating_count",
            ascending=False
        )[
            [
                "title",
                "genres",
                "rating_count",
                "average_rating"
            ]
        ].head(20).reset_index(drop=True)
    )

    print("\nRATING DISTRIBUTION")

    display(
        ratings["rating"].value_counts()
        .sort_index()
        .rename_axis("Rating")
        .reset_index(name="Number of Ratings")
    )

    plt.figure(figsize=(9, 5))

    sns.countplot(data=ratings, x="rating")

    plt.title("Distribution of Movie Ratings")
    plt.xlabel("Rating")
    plt.ylabel("Number of Ratings")
    plt.tight_layout()
    plt.show()


def evaluate_model():

    print("\nMODEL EVALUATION")

    if test_ratings.empty:

        print("Insufficient test data.")
        return

    evaluation_data = test_ratings.head(1000)

    global_mean = train_ratings["rating"].mean()

    actual = []
    baseline_predictions = []
    collaborative_predictions = []

    user_means = train_ratings.groupby(
        "userId"
    )["rating"].mean()

    movie_means = train_ratings.groupby(
        "movieId"
    )["rating"].mean()

    for _, row in evaluation_data.iterrows():

        user_id = int(row["userId"])
        movie_id = int(row["movieId"])
        actual_rating = float(row["rating"])

        baseline_predictions.append(global_mean)

        prediction = global_mean

        if (
            user_id in train_matrix.index
            and movie_id in train_matrix.columns
        ):

            similarities = user_similarity_df[user_id].drop(
                labels=[user_id],
                errors="ignore"
            )

            neighbor_ratings = train_matrix[movie_id]

            valid_users = neighbor_ratings[
                neighbor_ratings.notna()
            ].index

            similarities = similarities.reindex(
                valid_users
            ).dropna()

            similarities = similarities[
                similarities > 0
            ].sort_values(ascending=False).head(20)

            if not similarities.empty:

                values = train_matrix.loc[
                    similarities.index,
                    movie_id
                ]

                weights = similarities.reindex(
                    values.index
                ).fillna(0)

                if weights.sum() > 0:

                    prediction = np.average(
                        values.values,
                        weights=weights.values
                    )

        elif user_id in user_means.index:

            prediction = user_means[user_id]

        elif movie_id in movie_means.index:

            prediction = movie_means[movie_id]

        prediction = float(
            np.clip(prediction, 0.5, 5.0)
        )

        actual.append(actual_rating)
        collaborative_predictions.append(prediction)

    results = []

    for model_name, predictions in [
        ("Global Mean Baseline", baseline_predictions),
        (
            "User-Based Collaborative Filtering",
            collaborative_predictions
        )
    ]:

        mae = mean_absolute_error(actual, predictions)
        mse = mean_squared_error(actual, predictions)
        rmse = np.sqrt(mse)

        results.append({
            "Model": model_name,
            "MAE": round(mae, 4),
            "MSE": round(mse, 4),
            "RMSE": round(rmse, 4)
        })

    evaluation_results = pd.DataFrame(results)

    display(evaluation_results)

    evaluation_results.to_csv(
        os.path.join(OUTPUT_DIR, "model_evaluation.csv"),
        index=False
    )

    evaluation_results.set_index("Model")[
        ["MAE", "RMSE"]
    ].plot(
        kind="bar",
        figsize=(10, 5)
    )

    plt.title("Recommendation Model Evaluation")
    plt.ylabel("Error (lower is better)")
    plt.xticks(rotation=15, ha="right")
    plt.tight_layout()
    plt.show()

    print("Evaluation results saved.")


def save_project_outputs():

    print("\nSAVING PROJECT OUTPUTS")

    movies.to_csv(
        os.path.join(OUTPUT_DIR, "cleaned_movies.csv"),
        index=False
    )

    ratings.to_csv(
        os.path.join(OUTPUT_DIR, "cleaned_ratings.csv"),
        index=False
    )

    tags.to_csv(
        os.path.join(OUTPUT_DIR, "cleaned_tags.csv"),
        index=False
    )

    links.to_csv(
        os.path.join(OUTPUT_DIR, "cleaned_links.csv"),
        index=False
    )

    cleaning_report.to_csv(
        os.path.join(OUTPUT_DIR, "data_cleaning_report.csv"),
        index=False
    )

    missing_values_report.to_csv(
        os.path.join(OUTPUT_DIR, "missing_values_report.csv"),
        index=False
    )

    duplicate_report.to_csv(
        os.path.join(OUTPUT_DIR, "duplicate_report.csv"),
        index=False
    )

    summary.to_csv(
        os.path.join(OUTPUT_DIR, "project_summary.csv"),
        index=False
    )

    calculate_popular_movies().to_csv(
        os.path.join(OUTPUT_DIR, "popular_movies.csv"),
        index=False
    )

    genre_counts.rename_axis(
        "Genre"
    ).reset_index(
        name="Number of Movies"
    ).to_csv(
        os.path.join(OUTPUT_DIR, "genre_distribution.csv"),
        index=False
    )

    print("\nAll available project tables have been saved.")

    print("Output folder:", OUTPUT_DIR)

    print("\nFILES")

    for filename in sorted(os.listdir(OUTPUT_DIR)):
        print("-", filename)


# =====================================================================
# SECTION 11: MAIN MENU
# =====================================================================

def main_menu():

    while True:

        print("\n")
        print("=" * 65)
        print("                      MINEPTHEPY")
        print("             MOVIE RECOMMENDATION SYSTEM")
        print("=" * 65)

        print("1. Recommend Movies by Title")
        print("2. Recommend Movies for a User")
        print("3. Search Movies")
        print("4. View Popular Movies")
        print("5. Explore Movie Genres")
        print("6. View Dataset Statistics and Graphs")
        print("7. Evaluate Recommendation Model")
        print("8. Save Project Outputs")
        print("9. Exit")

        print("=" * 65)

        choice = input("Enter your choice (1-9): ").strip()

        try:

            if choice == "1":

                title = input(
                    "\nEnter a movie title: "
                ).strip()

                if title:
                    recommend_similar_movies(title, 10)
                else:
                    print("Please enter a movie title.")

            elif choice == "2":

                print(
                    "\nExample user IDs:",
                    sorted(ratings["userId"].unique())[:20]
                )

                user_id = input(
                    "Enter User ID: "
                ).strip()

                if user_id.isdigit():
                    recommend_for_user(int(user_id), 10)
                else:
                    print("Please enter a valid numeric User ID.")

            elif choice == "3":

                search_movies()

            elif choice == "4":

                show_popular_movies()

            elif choice == "5":

                explore_genres()

            elif choice == "6":

                show_dataset_statistics()

            elif choice == "7":

                evaluate_model()

            elif choice == "8":

                save_project_outputs()

            elif choice == "9":

                print("\nThank you for using MINEPTHEPY!")
                print("Movie Recommendation System closed.")
                break

            else:

                print("Invalid choice. Select a number from 1 to 9.")

        except KeyboardInterrupt:

            print("\nOperation cancelled.")

        except Exception as error:

            print("\nAn error occurred:", str(error))
            print("Please try another operation.")

        if choice != "9":

            input(
                "\nPress Enter to return to the main menu..."
            )


# =====================================================================
# SECTION 12: START THE PROJECT
# =====================================================================

print("\nInitializing the recommendation system...")

main_menu()
```
